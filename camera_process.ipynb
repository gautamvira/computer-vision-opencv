```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

Reading a video and creating an image from 4 frames

```python
cap = cv2.VideoCapture('assets/video.mp4')
ret, frame = cap.read()
frames = np.zeros(frame.shape, dtype=np.uint8)
height = frames.shape[0]
width = frames.shape[1]
cap.set(cv2.CAP_PROP_POS_FRAMES, 0)
pos = 0
for i in range(200):
    ret, frame = cap.read()
    if i % 50 == 0:
        if pos == 0:
            frames[:height//2, :width//2] = cv2.resize(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB), (0,0), fx=0.5, fy=0.5)
        elif pos == 1:
            frames[:height//2, width//2:width] = cv2.resize(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB), (0,0), fx=0.5, fy=0.5)
        elif pos == 2:
            frames[height//2:height, :width//2] = cv2.resize(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB), (0,0), fx=0.5, fy=0.5)
        elif pos == 3:
            frames[height//2:height, width//2:width] = cv2.resize(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB), (0,0), fx=0.5, fy=0.5)
        pos += 1
cap.release()
```

```python
fig, ax = plt.subplots(figsize=(10,10))
ax.imshow(frames)
ax.axis('off')
plt.show()
```
